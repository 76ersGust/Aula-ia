Exercício 19 - Crie dois arrays de numeros inteiros aleat8orios de tamanho 20 e calcule o
angulo entre eles (produto escalar normalizado).



> Usa produto escalar (np.dot) dividido pelas normas (np.linalg.norm) pra achar o cosseno do ângulo, depois converte pra graus com arccos + degrees



```python
import numpy as np

array1 = np.random.randint(1, 10, size=20)
array2 = np.random.randint(1, 10, size=20)
cos_theta = np.dot(array1, array2) / (np.linalg.norm(array1) * np.linalg.norm(array2))
angulo_graus = np.degrees(np.arccos(cos_theta)) #Ângulo entre dois vetores (produto escalar normalizado)
print(angulo_graus)
```

Exercício 16 - Resolva a equacao diferencial de segunda ordem d2y
dx2 +y = 0, com condicoes
iniciais y(0) = 1 e y(0) = 0.



> Transforma a equação de 2ª ordem num sistema de duas EDOs de 1ª ordem (y1=y, y2=y') pra poder usar solve_ivp



```python
from scipy.integrate import solve_ivp
import numpy as np

#Transformar em sistema de 1ª ordem: y1 = y, y2 = y'
def sistema(x, estado):
    y1, y2 = estado
    dy1_dx = y2
    dy2_dx = -y1
    return [dy1_dx, dy2_dx]

solucao = solve_ivp(sistema, [0, 10], [1, 0], t_eval=np.linspace(0, 10, 100))
print("x:", solucao.t)
print("y:", solucao.y[0])  # solução y(x) - essa é a função cosseno
```
