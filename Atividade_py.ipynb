EXERCICIO 4 - Converta uma temperatura fornecida em Celsius para Fahrenheit
usando a fórmula F = C * 9/5 + 32.

```python
celsius = float(input("Digite a temperatura em °C:"))
fahrenheit = (celsius * 9/5) + 32
print(f"{celsius}°C = {fahrenheit}°F")
```

EXERCICIO 5 - Crie um programa que imprima a tabuada de um número fornecido
pelo usuário.

```python
numero = int(input("Digite um número: "))
for i in range(1, 11):
    print(numero, "x", i, "=", numero * i)
```

EXERCICIO 8 - Crie um programa que verifique se uma palavra é um palíndromo
(ex: "radar").

"palavras iguais normal e de tras para frente"

```python
palavra = input("Digite uma palavra: ")
if palavra == palavra[::-1]:
    print("É palíndromo")
else:
    print("Não é palíndromo")
```
