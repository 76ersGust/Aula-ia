Exercício 19 - Crie dois arrays de numeros inteiros aleat8orios de tamanho 20 e calcule o
angulo entre eles (produto escalar normalizado).

```python
import numpy as np

array1 = np.random.randint(1, 10, size=20)
array2 = np.random.randint(1, 10, size=20)
cos_theta = np.dot(array1, array2) / (np.linalg.norm(array1) * np.linalg.norm(array2))
angulo_graus = np.degrees(np.arccos(cos_theta)) #Ângulo entre dois vetores (produto escalar normalizado)
```

Exercício 1 - Calcule a integral da funcao f(x) = x2 sin(x) no intervalo de 0 a π.

```python
import numpy as np
from scipy import integrate

resultado1, erro1 = integrate.quad(lambda x: x**2 * np.sin(x), 0, np.pi) # quando calcula a integral definida
print("Resultado:", resultado1, "| Erro:", erro1) #retorna (resultado, erro estimado)
```

Exercício 2 - Resolva a equacao diferencial dy
dx = x − y, com y(0) = 1, no intervalo de 0
a 10.

```python
import numpy as np
from scipy import integrate

sol2 = integrate.solve_ivp(lambda x, y: x - y, [0, 10], [1], t_eval=np.linspace(0, 10, 50)) #solve_ivp resolve equações diferenciais ordinárias a partir de uma condição inicial
print("x:", sol2.t)
print("y:", sol2.y[0])
```

Exercício 7 - Use a funcao de interpolacao cubica de SciPy para interpolar os dados
(0, 0), (1, 2), (2, 3), (3, 5), (4, 4).

```python
import numpy as np
from scipy import interpolate

x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 2, 3, 5, 4])
f = interpolate.interp1d(x, y, kind='cubic')  #kind='cubic' usa spline cúbica
print("Interpolado em x=1.5:", f(1.5))
```

Exercício 8 - Resolva o sistema de equa¸c˜oes 3x + 2y = 5 e x + 4y = 6.

Exercício 9 - Ajuste uma curva quadratica aos pontos (1, 2), (2, 4), (3, 5), (4, 7), (5, 8).

```python
import numpy as np
from scipy import optimize

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])
parametros9, _ = optimize.curve_fit(lambda x, a, b, c: a*x**2 + b*x + c, x, y) #curve_fit encontra os parâmetros a, b, c que melhor ajustam a função aos dados
print("Parâmetros [a, b, c]:", parametros9)
```

Exercício 10 - Calcule a integral dupla da funcao f(x, y) = xy + x2 no intervalo [0, 2] para x e [0, 1] para y.

```python
from scipy import integrate

resultado10, erro10 = integrate.dblquad(lambda y, x: x*y + x**2, 0, 2, lambda x: 0, lambda x: 1) #dblquad espera a função como f(y, x)
print("Integral dupla:", resultado10)
```
