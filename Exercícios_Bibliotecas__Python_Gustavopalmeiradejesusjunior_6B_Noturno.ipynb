Exercício 1 -

```python
import numpy as np
from scipy import integrate

integral, _ = integrate.quad(lambda x: (x**2) * np.sin(x), 0, np.pi)
```

Exercício 2 -

```python
import numpy as np
from scipy.integrate import solve_ivp

def dydx(x, y):
    return x - y

sol = solve_ivp(dydx, [0, 10], [1], t_eval=np.linspace(0, 10, 100))
```

Exercício 3 -

```python
from scipy.optimize import root_scalar

# Exemplo de busca de uma raiz com intervalo
raiz = root_scalar(lambda x: x**4 - 3*x**3 + 2, bracket=[0, 2])
```

Exercício 4 -

Exercício 5 -

```python
import numpy as np
from scipy.fft import fft

t = np.linspace(0, 1, 500)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)
fft_sinal = fft(sinal)
```

Exercício 16 -

```python
import numpy as np
from scipy.integrate import solve_ivp

# Sistema equivalente de 1ª ordem: y1' = y2, y2' = -y1
def sistema(x, y):
    return [y[1], -y[0]]

sol = solve_ivp(sistema, [0, 10], [1, 0], t_eval=np.linspace(0, 10, 100))
```
