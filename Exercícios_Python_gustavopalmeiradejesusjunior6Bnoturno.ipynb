EXERCICIO 4 - Converta uma temperatura fornecida em Celsius para Fahrenheit
usando a fórmula F = C * 9/5 + 32.

```python
celsius = float(input("Digite a temperatura em °C:"))
fahrenheit = (celsius * 9/5) + 32
print(f"{celsius}°C = {fahrenheit}°F")
```

(ex: "radar").

"palavras iguais da forma normal e de tras para frente"

EXERCICIO 12 - Conte o número de ocorrências de cada caractere em uma string
fornecida.

```python
texto = input("Digite um texto: ")
for caractere in set(texto):
    print(caractere, ":", texto.count(caractere))
```

EXERCICIO 26 - Crie uma calculadora que possa realizar adição, subtração,
multiplicação e divisão.

```python
a = float(input("Número 1: "))
op = input("Operação (+, -, *, /): ")
b = float(input("Número 2: "))
if op == '+': print(a + b)
elif op == '-': print(a - b)
elif op == '*': print(a * b)
elif op == '/': print(a / b)
```

EXERCICIO 27 - Verifique se duas palavras são anagramas (ou seja, se são formadas
pelas mesmas letras).

```python
palavra1 = input("Palavra 1: ")
palavra2 = input("Palavra 2: ")
if sorted(palavra1.lower()) == sorted(palavra2.lower()):
    print("São anagramas")
else:
    print("Não são anagramas")
```

EXERCICIO 34 - Conte o número de vogais em uma string fornecida.

```python
texto = input("Digite um texto: ").lower()
vogais = "aeiou"
total = sum(1 for char in texto if char in vogais)
print("Total de vogais:", total)
```
