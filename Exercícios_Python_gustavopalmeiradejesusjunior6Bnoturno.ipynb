Exercício 21 - Implemente a busca binária para encontrar um elemento em uma
lista ordenada.



> Divide a lista ao meio repetidamente até achar o elemento.

---





```python
def busca_binaria(lista, elemento):
    inicio, fim = 0, len(lista) - 1
    while inicio <= fim:
        meio = (inicio + fim) // 2
        if lista[meio] == elemento:
            return meio
        elif lista[meio] < elemento:
            inicio = meio + 1
        else:
            fim = meio - 1
    return -1
print(busca_binaria([10, 20, 30, 40, 50], 30))
```

Exercício 22 - Crie uma função para criptografar uma mensagem usando a cifra de
César.



> Desloca cada letra um número fixo de posições no alfabeto.

---





```python
def cifrar(texto, chave):
    resultado = ""
    for c in texto:
        resultado += chr(ord(c) + chave) if c.isalpha() else c
    return resultado
print(cifrar("python", 3))
```

Exercício 24 - Implemente uma função para validar um CPF brasileiro.



> Confere os dois dígitos verificadores do CPF ultilizando a fórmula oficial.

---





```python
def validar_cpf(cpf):
    cpf = ''.join(filter(str.isdigit, cpf))
    if len(cpf) != 11:
        return False
    if cpf == cpf[0] * 11:
        return False
    soma = sum(int(cpf[i]) * (10 - i) for i in range(9))
    resto = (soma * 10) % 11
    primeiro_digito = 0 if resto == 10 else resto
    if primeiro_digito != int(cpf[9]):
        return False

    soma = sum(int(cpf[i]) * (11 - i) for i in range(10))
    resto = (soma * 10) % 11
    segundo_digito = 0 if resto == 10 else resto
    if segundo_digito != int(cpf[10]):
        return False
    return True
cpf = input("Digite o CPF (com ou sem pontuação): ")
if validar_cpf(cpf):
    print("CPF válido")
else:
    print("CPF inválido")
```

Exercício 31 - Implemente o algoritmo de ordenação por bolha (Bubble Sort).



> Compara pares vizinhos e troca se estiverem fora de ordem, repetindo várias vezes.

---





```python
lista = [64, 34, 25, 12, 22, 11]
for i in range(len(lista)):
    for j in range(0, len(lista) - i - 1):
        if lista[j] > lista[j + 1]:
            lista[j], lista[j + 1] = lista[j + 1], lista[j]
print("Lista ordenada:", lista)
```

> A função chama ela mesma até chegar em 1.

---



```python
def fatorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * fatorial(n - 1)
numero = int(input("Digite um número: "))
resultado = fatorial(numero)
print(f"O fatorial de {numero} é {resultado}")
```

Exercício 48 - Implemente o algoritmo de ordenação por inserção (Insertion Sort).



> Pega cada número e o "encaixa" na posição certa entre os já ordenados

---






```python
lista = [12, 11, 13, 5, 6, 77, 22]
for i in range(1, len(lista)):
    chave = lista[i]
    j = i - 1
    while j >= 0 and chave < lista[j]:
        lista[j + 1] = lista[j]
        j -= 1
    lista[j + 1] = chave
print("Ordenada:", lista)
```
